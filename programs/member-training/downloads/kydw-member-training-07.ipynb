## 图表数据与来源

同一组样本在两个方案里必须对应同一行。下面使用明确标为人工的指标表，配对连线对应每个样本在两个方案中的分数。

```python
example = WORK/'illustrative_metrics.csv'
rows = [
    {'image':'demo01','reference_mae':0.14,'model_mae':0.11},
    {'image':'demo02','reference_mae':0.09,'model_mae':0.10},
    {'image':'demo03','reference_mae':0.18,'model_mae':0.13},
    {'image':'demo04','reference_mae':0.12,'model_mae':0.11},
]
with example.open('w',newline='',encoding='utf-8') as f:
    writer=csv.DictWriter(f,fieldnames=list(rows[0]));writer.writeheader();writer.writerows(rows)
with example.open(encoding='utf-8') as f:
    loaded=list(csv.DictReader(f))
scores=torch.tensor([[float(r['reference_mae']),float(r['model_mae'])] for r in loaded])
print('illustrative means:',scores.mean(dim=0))
```

```python
fig,ax=plt.subplots(figsize=(6,4))
for i,row in enumerate(scores):
    ax.plot([0,1],row,marker='o',alpha=0.7,label=loaded[i]['image'])
ax.set_xticks([0,1],['Reference','Model'])
ax.set_ylabel('RGB MAE (0-1)')
ax.set_title('Illustrative data — not experiment results')
ax.legend()
fig.tight_layout()
fig.savefig(WORK/'illustrative_comparison.png',dpi=180)
plt.show()
```
