## 同一照片的 R、G、B 层

单独显示一个颜色分量，会保留原照片的空间排列。三个通道都记录同一组空间位置。把 `CHANNEL` 改成 0、1、2，查看选中的层；彩色层以对应颜色显示，数值没有变成另一张照片。

```python
CHANNEL=0  # 0 红、1 绿、2 蓝
layers=[]
for channel in range(3):
    layer=np.zeros_like(cat_array)
    layer[:,:,channel]=cat_array[:,:,channel]  # 保留一个通道，其余分量归零
    layers.append(layer)
view([cat_array,*layers],['原照片','R 红层','G 绿层','B 蓝层'])
plt.figure(figsize=(5,3));plt.imshow(cat_array[:,:,CHANNEL],cmap='gray',vmin=0,vmax=255);plt.title('选中通道的强度视图');plt.axis('off');plt.show()
print('同一像素三个通道:',cat_array[110,350].tolist())
```

## RGBA 与背景

透明度为 0.5 时，显示颜色由前景和背景共同决定。下面按 Alpha 混合算式计算白底与黑底上的颜色。把 `ALPHA` 改成 0、0.5、1，查看同一个前景为什么会显示得不同。

```python
ALPHA=0.5  # 0 完全透明，1 完全不透明
foreground=np.array([245,160,40],dtype=np.float32)
white=ALPHA*foreground+(1-ALPHA)*255
black=ALPHA*foreground+(1-ALPHA)*0
view([np.tile(white/255,(40,40,1)),np.tile(black/255,(40,40,1))],['白色背景上的显示','黑色背景上的显示'])
print('RGBA:',foreground.tolist()+[ALPHA],'白底 RGB:',white,'黑底 RGB:',black)
```

## BGR 存储次序

同一颜色的红、绿、蓝可以按另一种次序保存。把 BGR 数组直接交给 RGB 显示函数，会交换红蓝。下面分别显示错误解释和恢复后的结果，核对恢复是否与原图相同。

```python
bgr=cat_array[:,:,::-1].copy()  # RGB 改为 BGR 存储顺序
restored=bgr[:,:,::-1]
view([cat_array,bgr,restored],['原 RGB','BGR 误作 RGB 显示','交换回 RGB'])
assert np.array_equal(restored,cat_array)
print('原位置 RGB:',cat_array[110,350],'BGR 存储:',bgr[110,350])
```

## 数值类型与内存

同样尺寸的图像有相同数量的颜色数值，但每个数占用的字节数不同。256×256×3 个数占用的内存可以通过数组的 `nbytes` 核对。把边长改为 512，看看内存是否变成四倍。

```python
SIDE=256  # 可改成 512
uint_image=np.zeros((SIDE,SIDE,3),dtype=np.uint8)
float_image=uint_image.astype(np.float32)
print('数值数量:',uint_image.size)
print('uint8:',uint_image.nbytes/1024,'KiB；float32:',float_image.nbytes/1024,'KiB')
fig,ax=plt.subplots(figsize=(6,3));ax.bar(['uint8','float32'],[uint_image.nbytes/1024,float_image.nbytes/1024],color=['#34878c','#e6a95f']);ax.set_ylabel('KiB · 仅数组数据');plt.show()
fig,ax=plt.subplots(figsize=(8,3));values=[uint_image.nbytes/1024,float_image.nbytes/1024]
ax.barh(['uint8','float32'],values,color=['#68a296','#315e75']);ax.set_xlim(0,max(values)*1.23);ax.set_xlabel('KiB (1 KiB = 1024 bytes)')
for i,v in enumerate(values):ax.text(v+max(values)*.02,i,f'{v:g} KiB',va='center')
plt.tight_layout();plt.show();plt.close(fig)
```

猫照片缩成少量像素后，局部细节会消失；放大可以看清留下的网格。最近邻保留方块，双线性根据邻近数值计算过渡。改变 `SIDE`，观察能保留的结构；放大没有重新拍摄已经丢掉的细节。

## 范围缩放与直方图

把原始数值除以 255 或映射到 −1—1 后，直方图的横轴会改变。显示前按相应规则恢复，照片仍保持原来的颜色。直方图显示的是计算数值，不是为了好看而重新拉伸的颜色。

```python
unit=cat_array.astype(np.float32)/255
signed=2*unit-1
variants=[('uint8',cat_array,cat_array),('/255',unit,unit),('[-1,1]',signed,(signed+1)/2)]
fig,axes=plt.subplots(2,3,figsize=(12,6))
for i,(name,data,shown) in enumerate(variants):
    axes[0,i].imshow(shown);axes[0,i].set_title(name+' · 显示已恢复');axes[0,i].axis('off')
    axes[1,i].hist(data[:,:,0].ravel(),bins=32,color='#cc685e');axes[1,i].set(xlabel='实际 R 数值',ylabel='像素数')
plt.tight_layout();plt.show()
print('同一位置:',cat_array[110,350],unit[110,350],signed[110,350])
```

## Min–Max 与标准化

这次我们用当前照片的通道极值、均值和标准差作演示。Min–Max 把极值拉到 0 和 1；标准化减去均值再除以标准差。观察直方图，然后反算标准化数值，核对颜色是否恢复。实际训练应保存训练集统计量，并让验证和测试使用相同参数。

```python
minimum=unit.min(axis=(0,1),keepdims=True);maximum=unit.max(axis=(0,1),keepdims=True)
mean=unit.mean(axis=(0,1),keepdims=True,dtype=np.float64);std=unit.std(axis=(0,1),keepdims=True,dtype=np.float64)
minmax=(unit-minimum)/np.maximum(maximum-minimum,1e-8)  # 处理常数通道的零分母
standard=(unit-mean)/np.maximum(std,1e-8)
recovered=standard*std+mean  # 显示前还原，保留计算张量的独立含义
fig,axes=plt.subplots(2,3,figsize=(12,6))
for i,(name,data,shown) in enumerate([('原图',unit,unit),('Min–Max',minmax,minmax),('标准化后还原显示',standard,recovered)]):
    axes[0,i].imshow(np.clip(shown,0,1));axes[0,i].set_title(name);axes[0,i].axis('off')
    axes[1,i].hist(data[:,:,0].ravel(),bins=32,color='#cc685e');axes[1,i].set(xlabel='实际 R 数值',ylabel='像素数')
plt.tight_layout();plt.show()
print('通道均值:',mean.flatten(),'标准差:',std.flatten(),'标准化后的均值:',standard.mean((0,1)))
assert np.allclose(recovered,unit,atol=1e-6)
```

## 零维至五维张量

一个温度值、沿时间排列的读数、灰度图的行列，都可以用张量保存。维数是在数轴，不是在数元素。下面用真实照片组成前三种图像对象；视频批量只用空数组说明排列，没有把它称作真实视频。

```python
objects={
    '0D scalar · 一个示例温度':torch.tensor(23.5),
    '1D vector · 三个 RGB 数':torch.tensor([.2,.5,.8]),
    '2D matrix · 灰度图片':torch.from_numpy(gray_photo.copy()),
    '3D tensor · RGB 图片 CHW':torch.from_numpy(cat_array.copy()).permute(2,0,1),
    '4D tensor · 图片批量 BCHW':torch.stack([photo('pets/'+s['image']) for s in samples]),
    '5D tensor · 视频批量排列示意 BTCHW':torch.zeros(2,4,3,16,16),
}
for name,value in objects.items():print(name,'shape=',list(value.shape),'轴数=',value.ndim,'元素数=',value.numel())
print('一个含 1000 个数的向量仍为一维：',torch.arange(1000).ndim)
view([gray_photo,cat_array],['2D 行、列','3D 颜色、行、列'])
```

## 轴顺序与索引

行、列、通道排列的照片转为通道、行、列后，同一位置的颜色应当保持一致。用索引核对这一点，再比较只改形状和真正交换轴的差别。

```python
hwc=torch.from_numpy(cat_array.copy())
chw=hwc.permute(2,0,1)  # 交换轴的顺序
ROW,COL=110,350
print('HWC:',hwc.shape,'CHW:',chw.shape)
print('同一位置:',hwc[ROW,COL,:],chw[:,ROW,COL])
assert torch.equal(hwc[ROW,COL,:],chw[:,ROW,COL])
wrong=hwc.reshape(3,hwc.shape[0],hwc.shape[1])  # 只改变分组方式，不等于交换轴
view([chw.float()/255,wrong.float()/255],['正确 permute','reshape 误当轴交换'])
```

## 批量与对应标签

每张图片都有独立编号，标签必须跟着相同编号移动。原顺序和重排后的顺序使用同一组照片与标签。改动 `ORDER`，观察每列照片和名称仍对应。

```python
batch=torch.stack([photo('pets/'+s['image'],96) for s in samples])
labels=torch.tensor([s['species']-1 for s in samples])
ORDER=[2,0,3,1]  # 改动排列时让图像、标签一起采用这个顺序
view([batch[i] for i in ORDER],['cat 0' if labels[i]==0 else 'dog 1' for i in ORDER])
print('批量形状:',list(batch.shape),'标签形状:',list(labels.shape),'新标签次序:',labels[ORDER].tolist())
assert batch.shape[0]==labels.numel()
```

## 按通道汇总与广播

每层颜色计算一个均值，需要对行、列作汇总，保留通道。减去均值时给行、列留出单独的轴，才能让三个数分别对应三个颜色层。

```python
channel_mean=batch.mean(dim=(2,3))  # [B,C]
centered=batch-channel_mean[:,:,None,None]  # 每张图、每个通道分别减去均值
print('通道均值形状:',list(channel_mean.shape),'总体均值形状:',list(batch.mean((1,2,3)).shape))
print('中心化后通道均值:',centered.mean((2,3)))
assert torch.allclose(centered.mean((2,3)),torch.zeros_like(channel_mean),atol=1e-6)
```
