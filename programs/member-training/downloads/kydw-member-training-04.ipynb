## 空间尺寸

照片的每个位置都需要得到一个分数。小网络先缩小特征网格，再恢复到输入尺寸；对照照片和标注检查形状。

```python
item = next(item for item in samples if item.get('mask'))
image = photo('pets/' + item['image'])
raw = read_image(str(fetch('pets/' + item['mask'])), mode=ImageReadMode.UNCHANGED)
assert raw.shape[0] == 1, '需要原始单通道 trimap，不能使用着色后的展示图片。'
trimap = F.interpolate(raw[None].float(), size=(64,64), mode='nearest')[0,0].long()
valid = trimap != 3  # 边界待定区不参与损失与评价
target = trimap == 1  # 宠物前景为 True；原始背景编码是 2
assert set(trimap.unique().tolist()).issubset({1,2,3})
```

```python
class TinySegmenter(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(nn.Conv2d(3,8,3,padding=1),nn.ReLU(),nn.MaxPool2d(2))  # 空间边长减半
        self.head = nn.Conv2d(8,1,1)  # 每个位置输出一个前景分数
    def forward(self,x):
        features = self.encoder(x)
        restored = F.interpolate(features,size=x.shape[-2:],mode='bilinear',align_corners=False)  # 恢复输入尺寸
        return self.head(restored)

model = TinySegmenter()
scores = model(image[None])
print('input:',image[None].shape,'output:',scores.shape)
assert scores.shape == (1,1,64,64)
```

## 固定起点的一步比较

两个学习率若从同一组参数出发，第一次更新前的损失应相同。下面只更新一步，查看参数变化，不把它当成完整测试性能。

```python
initial = {name:value.detach().clone() for name,value in model.state_dict().items()}
observations = []
for learning_rate in [0.01,0.1]:
    candidate = TinySegmenter()
    candidate.load_state_dict(initial)  # 两个候选模型从完全相同的参数开始
    optimizer = torch.optim.SGD(candidate.parameters(),lr=learning_rate)
    optimizer.zero_grad()
    pixel_loss = F.binary_cross_entropy_with_logits(candidate(image[None])[0,0],target.float(),reduction='none')
    loss = pixel_loss[valid].mean()  # 仅使用有效像素
    loss.backward()
    optimizer.step()  # 此处每个设置只更新一次
    change = sum((value-initial[name]).abs().sum() for name,value in candidate.state_dict().items())
    observations.append({'learning_rate':learning_rate,'initial_loss':loss.item(),'parameter_change':change.item()})
print(observations)
assert abs(observations[0]['initial_loss']-observations[1]['initial_loss']) < 1e-6
```

### 同一步更新的两项观察

两组开始时的损失相同，参数变化量却会随学习率改变。把两个学习率设成相同值，重新运行比较代码和下图；此处横轴是学习率，柱高只表示一步更新量。

```python
plt.figure(figsize=(6,3))
plt.bar([str(row['learning_rate']) for row in observations],[row['parameter_change'] for row in observations],color=['#358b8e','#dfab69'])
plt.xlabel('本次比较的学习率');plt.ylabel('一步参数绝对变化量');plt.title('实际单步计算 · 不是测试性能');plt.show()
```
